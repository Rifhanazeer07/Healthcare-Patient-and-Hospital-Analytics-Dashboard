```python
import pandas as pd
```

```python
data=pd.read_csv('patients.csv')
data.head(20)
```

```python
data['patient_id'].duplicated().sum()
```

```python
data.isnull().sum()
```

```python
data['insurance_type']=data['insurance_type'].fillna('Unknown')
data.isnull().sum()
```

```python
data['gender'] = data['gender'].replace({
    'M': 'Male',
    'F': 'Female'
})
data['gender'].value_counts()
```

```python
print(data['age'].describe())
data[data['age'] < 0]
data[data['age'] > 120]
```

```python
data['age_group'] = pd.cut(
    data['age'],
    bins=[-1, 17, 35, 50, 65, 120],
    labels=['Child', 'Young Adult', 'Adult', 'Middle-aged', 'Senior']
)
data['age_group'].value_counts()
```

```python
data[['age', 'age_group']].head(20)
```

```python
data['comorbidity_count'].describe()
data[data['comorbidity_count'] < 0]
```

```python
data['prev_admissions'].describe()
data[data['prev_admissions'] < 0]
```

```python
data.info()
```

```python
data.to_csv("patients_cleaned.csv", index=False)
```
